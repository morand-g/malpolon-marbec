```python
import numpy as np
import pandas as pd 
from pathlib import Path



import plotly
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go


from tqdm.auto import tqdm
tqdm.pandas()

THRESHOLD = 0.226
```

```python
inputpath = Path('/marbec-data/RLS-Australia/malpolon/inputs/australiapreds')
predpath = Path('/marbec-data/RLS-Australia/malpolon/outputs/37_mm4_transductif-2026-02-20_14-48/predictions-probs.csv')

pa_predictions = pd.read_csv(predpath, index_col='survey_id')
metadata = pd.read_csv(inputpath / 'auspredgrid.csv', index_col='survey_id', usecols=['survey_id', 'depth', 'latitude', 'longitude', 'eventDate'])
```

## UICN

```python
species = ['Hypoplectrodes maccullochi', 'Kyphosus cornelii', 'Parma unifasciata', 'Mecaenichthys immaculatus']
uicnpreds= pd.merge(left=metadata, right=pa_predictions[species], left_index=True, right_index=True)
```

```python
(uicnpreds[species] > THRESHOLD).sum()
```

```python
fig = px.scatter_map(
    uicnpreds,
    lat="latitude",
    lon="longitude",
    #color='Hypoplectrodes maccullochi',
    color = 'Kyphosus cornelii',
    center={'lat':-28, 'lon':133}, zoom=3.5,
    height = 1100, width = 1200, map_style='light',
    color_continuous_scale=px.colors.sequential.Inferno,
)


fig.show()
```
